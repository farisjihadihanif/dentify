## Conclusion

On 255 identical held-out test images, CLAHE preprocessing (Variant B) achieved 81.57% accuracy (Macro F1 0.7784) vs Baseline (Variant A) at 67.84% accuracy (Macro F1 0.6608), a delta of +13.73 percentage points. Discordant pairs: A-only=12, B-only=47. Exact McNemar p-value=0.0000 (statistically significant at alpha=0.05; Variant B accuracy is higher than Variant A).
